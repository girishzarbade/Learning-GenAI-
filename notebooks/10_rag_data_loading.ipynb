### Text Document Loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/long-doc.txt")
texts = loader.load()
len(texts)
oneDoc = texts[0]
print(oneDoc.page_content)
```

### PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("../data/sample-report.pdf")
docs = loader.load()
len(docs)
```

```python
oneDoc = docs[0]
print(oneDoc.page_content)
```

### CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("../data/mixed-types.csv")
docs = loader.load()
len(docs)

```

### Web Page Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path="https://docs.langchain.com/oss/python/integrations/document_loaders")
docs = loader.load()
```

```python
docs[0].metadata
print(docs[0].page_content)
```

### Wikipedia Loader
