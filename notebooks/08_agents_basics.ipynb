```python
from dotenv import load_dotenv  
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers (a:int , b:int):
    """
        It will return sum of two numbers
        
    Args:
        a (int):Number One
        b (int): Number Two
    """
    
    return a + b
```

```python
add_numbers.invoke({"a":14, "b":15})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm = ChatOpenAI(model = "gpt-4")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers],
    system_prompt="You will act as a calculator, and will be used for calculation"
)
```

```python
response = agent.invoke({"messages":[{"role":"user", "content":"what is 4 + 12?"}]})
```

```python
response['messages']
```

To understand the above response we can break it down into individual lists by applying loop 

```python
for res in response['messages']:
    print(res)
    print("\n")
```
