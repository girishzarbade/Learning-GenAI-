```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b", streaming=True)
```

```python
question = "Explain GenAI"
```

```python
response = llm.stream(question)
```

```python
for chunk in response:
    print(chunk, "\n" )
```
