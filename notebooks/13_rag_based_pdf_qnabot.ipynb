```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.prompts import PromptTemplate
```

```python
loader = PyPDFLoader("../data/Frontend-Roadmap.pdf")
docs = loader.load()
len(docs)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size= 1000,chunk_overlap= 200)
splitted_data = splitter.split_documents(docs)
len(splitted_data)
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-large") 
```

```python
vector_store =Chroma.from_documents(
    documents =splitted_data,
    embedding = embeddings
)
```

```python
query = "React Content"
data = vector_store.similarity_search(query=query)
```

```python
data[0]
```

```python
context = ""
for doc in data:
    context += doc.page_content + "\n"
```

```python
llm = ChatOpenAI(model="gpt-5")
```

```python
### Chain - Context_generate | prompt | llm | strparser
```

```python
def get_context(query:str):
    data = vector_store.similarity_search(query=query)
    context = ""
    for doc in data:
        context += doc.page_content + "\n"
        
    return{
        "context":context,
        "question": query
    }
```

```python
prompt = PromptTemplate.from_template(""" You are a helpful assistant and provide answerd based on the template if you  don't know the answer, then you can say I don't know
                                      context:{context}  
                                      question:{question}""")
```

```python
rag_chain =get_context | prompt | llm
```

```python
res = rag_chain.invoke("What are the topics in typescript?")
```
