```python
from pydantic import BaseModel
from langgraph.graph import StateGraph, START, END
```

```python
class GreetState(BaseModel):
    message: str =" "
```

```python
graph = StateGraph(GreetState)
```

```python
### Nodes - Python Functions
```

```python
def Greet(state:GreetState):
    state.message = f"{state.message}! All Good!"
    return state 


def upperCase(state:GreetState):
    state.message = state.message.upper()
    return state
```

```python
graph.add_node("FirstNode",Greet)
```

```python
graph.add_node("uppe_case", upperCase)
```

```python
graph.add_edge(START, "FirstNode")
graph.add_edge("FirstNode", "upperCase")
graph.add_edge("FirstNode", "upperCase")
graph.add_edge("upperCase", END)
```
